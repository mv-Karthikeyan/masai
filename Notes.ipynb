```python
# n even numbers:    here end number is 2(n-1),2n-2 . but in range it will take till the before number. So 2n-1 is end number.
n=int(input("enter n number:"))
if n>=2:
  for i in range(2,2*n-1,2):
    print(i)
else:
  print("give correct number")
```
