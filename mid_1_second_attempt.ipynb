```python
import pandas as pd
import numpy as np

np.random.seed(45)

sales_df={
    'region':np.random.choice(['North','South','East','West'],size=10),
    'products':np.random.choice(['A','B','C'],size=10),
    'units_sold':np.random.randint(1,100,size=10),

    'revenue':[100.0,200.0,300.0,400.0,500.0,111.0,333.0,111.0,454.0,221.0]


}
sales_df=pd.DataFrame(sales_df)
print(f"sales_df:\n{sales_df}\n")
print("*"*100)

summary=[
    sales_df.groupby(['region','products']).agg(
        total_units_sold=('units_sold','sum'),
        average_revenue=('revenue','mean'),
        Total_count=('units_sold','count')



    ).reset_index()

]
print(f"summary of the Dataframe:\n{summary}")
```
