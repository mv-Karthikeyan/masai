```python
def num_sum(num):
  sum=0
  for i in range(len(num)):
    if num[i]%2==0:
      sum+=num[i]
  return sum
num=list(map(int,input("Enter the list of numbers:").split()))
result=num_sum(num)
print(f"The sum of even numbers:{result}")
```

```python
def num_sum(num,operation):
  sum=0
  for i in range(len(num)):
    if operation and num[i]%2==0:
      sum+=num[i]
    elif not operation and num[i]%2!=0:
      sum+=num[i]
  return sum
num=list(map(int,input("Enter the list of numbers:").split()))
result=num_sum(num,True)

print(f"the value is:{result}")
```

```python
#check anagram using functions. Anagram means if sorted of word1 is equal to the sorted of other word then they both are anagram

def anagram():
  word1=input("please enter the word 1:").lower()
  word2=input("please enter the word 2:").lower()
  x=sorted(word1)
  y=sorted(word2)
  return x==y
anagram()

```

```python
def calculate_total(marks):
  sum=0
  for i in range(len(marks)):
    sum+=marks[i]
  return sum
marks=[88,76,95,60,82]
result=calculate_total(marks)
print(f"Total Marks = {result}")

```

```python
#Task:2
def calculate_average(marks):
  avg=calculate_total(marks)/len(marks)
  return avg
Average=calculate_average(marks)
print(f"Average of the marks={Average}")
```
