```python
# Takes a number as input
# If the number is positive (> 0), print "Positive number"
# If the number is negative (< 0), print "Negative number"
# If the number is zero (== 0), print "Zero"

num=int(input("please enter a number:"))
if num==0:
  print(f"Your num {num} is Zero")
elif num>0:
  print(f"Your num {num} is Positive number")
elif num<0:
  print(f"Your num {num} is Negative number")
```

```python
# Takes user's age as input
# If age is 18 or above, print "You can vote"
# If age is 21 or above, print "You can drink alcohol (in US)"
# Always print "Thank you" at the end
age=int(input("Please enter your age:"))
if age>=18:
  print(f"Your age is {age},So you can vote")
if age>=21:
  print(f"Your age is {age},So you can drink alcohol (in US)")
elif age>18 and age<21:
  print(f"and You are not allowed drink alcohol")
else:
  print(f"Your age is {age},So you cannot drink alcohol (in US) and you are not allowed to Vote")
print("Thankyou")
```

```python
# Take a score (0-100) as input
# Use multiple if statements to:
# Check if score >= 90: print "Grade: A" and "Excellent work!"
# Check if score >= 80 and < 90: print "Grade: B" and "Good job!"
# Check if score >= 70 and < 80: print "Grade: C" and "Satisfactory"
# Check if score >= 60 and < 70: print "Grade: D" and "Needs improvement"
# Check if score < 60: print "Grade: F" and "Failed"
# Also check:
# If score == 100: print "Perfect score!"
# If score < 50: print "Please see instructor"


score=int(input("Please enter your Score:"))
if score>=90 and score<100:
  print(f"Your score is {score},You are in Grade:A , Excellent work!")
elif score>=80 and score<90:
  print(f"Your score is {score},You are in Grade:B , Good Job!")
elif score>=70 and score<80:
  print(f"Your score is {score},You are in Grade:C , Satisfactory")
elif score>=60 and score<70:
  print(f"Your score is {score},You are in Grade:D , Needs improvement")
elif score<50:
  print("Please see the instructor")
elif score<60:
  print(f"Your score is {score},You are in Grade:F , Failed")

elif score==100:
  print(f"Perfect Score!")
```
